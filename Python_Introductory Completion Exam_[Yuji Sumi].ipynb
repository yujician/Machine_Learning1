Q.04

処理：引数にタプルを与えると全ての要素を３倍したリストを返却する  
例：(1,2,3)→[3,6,9]

```python
def q04(tpl):
    ret = []
    for i in tpl:
        ret.append(i*3)
    return ret

tpl = (1,2,3)
q04(tpl)
```

Q.10    
引数：なし  
処理：以下の計算結果をnumpyの配列で返却する   
$\displaystyle 3\left(\begin{array}{ccc} 1 & 2 & 3\\ 4 & 5 & 6 \\ \end{array}\right) - 2\left(\begin{array}{ccc} 1 & 2 & 3\\ 1 & 2 & 3 \\ \end{array}\right)$

```python
def q10():
    array_1 = np.arange(1,7).reshape(2,3)
    array_2 = np.array([1,2,3])
    array_3 = 3*array_1 - 2*array_2
    return array_3

q10()
```
